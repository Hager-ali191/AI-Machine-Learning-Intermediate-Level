### Data Simulation
Kwame has set up a simulated data block here so development can continue seamlessly even if the live database connection is temporarily dropped. This DataFrame precisely mirrors the structure of the `regional_sales_data.csv` extract.

```python
# Simulated live-data generation loop (mimics regional_sales_data.csv)
np.random.seed(42)
months = ['January', 'February', 'March', 'April', 'May', 'June']
regions = ['North America', 'EMEA', 'APAC', 'LATAM']

data = []
for month in months:
    for region in regions:
        ad_spend = np.random.randint(15000, 75000)
        conversions = int((ad_spend / 100) * np.random.uniform(0.3, 1.8))
        revenue = conversions * np.random.uniform(150, 300)
        data.append([month, region, ad_spend, conversions, revenue])

df = pd.DataFrame(data, columns=['Month', 'Region', 'Ad Spend', 'Conversions', 'Revenue'])

# Display the first few rows to verify the structure
display(df.head())
```
